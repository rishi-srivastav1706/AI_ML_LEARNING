```python
import pandas as pd
```

```python
info = {
    "name" : ["Rahul", "Rishi", "Riddhi"],
    "CGPA":[9.5,7.5,8.5]
}

df = pd.DataFrame(info)

print(df)
```

```python
s = pd.Series([1,2,3,4,5])
print(s)

#index
print(s[0])
print(s[2])
```

```python
s = pd.Series([23,24,33,42,"abc"], index= ["Adam", "Eve", "Bob", "Charlie", 25])
print(s)

print(s["Adam"])
print(s["Bob"])

print(s.index)
```

```python
s1 = pd.Series([1,2,3,4,5])
s2 = pd.Series([10,20,30,40,50])

print(s1+s2)

s1[0] = 100
print(s1)
changed_s1 = s1.drop(0)
print(s1)
print(changed_s1)
```

```python
# dataframe

info = {
    "Name" :["Adam", "Eve","Bob"],
    "Age" :[23,24,25],
    "GPA" :[9.5,8.6,7.2]
}
df = pd.DataFrame(info)

print(df, type(df))

print(df.index)
print(df.columns)
```

```python
df = pd.DataFrame([["Adam", 23], ["Bob" , 23], ["Eve", 23]], columns=["Name", "age"])
print(df)
```
